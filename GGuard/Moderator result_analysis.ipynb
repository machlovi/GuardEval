```python
import os
import pandas as pd
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

input_directory = "/home/naseem_fordham/LLM_evaluation/Moderators/Chatbase Moderators/Catplus_results/Metahate_Moderators/"
output_file = "Wild_metrics_results.csv"

# Only include CSV files that contain "AegisGuard" in their name
csv_files = [
    f for f in os.listdir(input_directory)
    if f.endswith(".csv") and "Wild" in f
]

column_pairs = [
    ('prompt_label', 'prompt_predic_label'),
    ('response_label', 'response_predic_label')
]

all_results = []

for file in csv_files:
    file_path = os.path.join(input_directory, file)
    
    try:
        df = pd.read_csv(file_path)
        
        for true_col, pred_col in column_pairs:
            # Skip if either column doesn't exist
            if true_col not in df.columns or pred_col not in df.columns:
                print(f"Skipping {file}: Missing {true_col} or {pred_col}")
                continue
            
            # Skip if both columns are empty
            if df[true_col].dropna().empty or df[pred_col].dropna().empty:
                print(f"Skipping {file}: {true_col} or {pred_col} is empty")
                continue
            
            # Process and standardize label texts
            true_labels = df[true_col].fillna("").astype(str).str.lower()
            pred_labels = df[pred_col].fillna("").astype(str).str.lower()
            
            # Calculate accuracy
            accuracy = accuracy_score(true_labels, pred_labels)
            
            # Only consider valid classes
            available_labels = list(set(true_labels.unique()) & {"safe", "unsafe"})
            if not available_labels:
                print(f"Skipping {file}: No valid labels found in {true_col}")
                continue
            
            # Calculate per-class metrics
            precision, recall, f1, support = precision_recall_fscore_support(
                true_labels,
                pred_labels,
                labels=available_labels,
                zero_division=0
            )
            
            # Macro-average metrics
            macro_precision = precision.mean()
            macro_recall = recall.mean()
            macro_f1 = f1.mean()
            
            # Save per-class results
            for i, label in enumerate(available_labels):
                all_results.append({
                    "File": file,
                    "Label Type": true_col,
                    "Class": label,
                    "Accuracy": accuracy,
                    "Precision": precision[i],
                    "Recall": recall[i],
                    "F1-score": f1[i],
                    "Support": support[i],
                    "Average": "per-class",
                })
            
            # Save macro-average result
            all_results.append({
                "File": file,
                "Label Type": true_col,
                "Class": "macro_avg",
                "Accuracy": accuracy,
                "Precision": macro_precision,
                "Recall": macro_recall,
                "F1-score": macro_f1,
                "Support": support.sum(),
                "Average": "macro",
            })
    
    except Exception as e:
        print(f"Error processing {file}: {e}")

# Save the results
metrics_df = pd.DataFrame(all_results)
metrics_df.to_csv(output_file, index=False)
print(f"Metrics saved to {output_file}")
```

```python
import os
import pandas as pd
# df_gptft=pd.read_csv('/home/naseem_fordham/LLM_evaluation/HateCheck/GPTHC_SafeGemma_FinetuneGPTHC_FS/SafeGemma_GPHTC_FT.csv')
input_dir = '/home/naseem_fordham/LLM_evaluation/HateCheck'
all_metrics = []

for file_name in os.listdir(input_dir):
    file_path = os.path.join(input_dir, file_name)
    if file_name.endswith('.csv') and file_name.startswith('GPTHC_'):
        print(f"Processing file: {file_name}")
        df = pd.read_csv(file_path)
        df['source_file'] = file_name.split('.')[0] 
        all_metrics.append(df)

# Concatenate all DataFrames
combined_df = pd.concat(all_metrics, ignore_index=True)
# combined_df=pd.concat([combined_df,df_gptft],axis=0)



# Display the combined DataFrame
combined_df
```

```python
combined_df['source_file'].unique()

model_mapping = {
    'GPTHC_wildgaurd': 'WildGuard',
    'GPTHC_OAI': 'OpenAI',
    'GPTHC_llama': 'LlamaGuard',
    'GPTHC_SafeGemma': 'SafeGemma',
    'GPTHC_ShieldGemma':'ShieldGemma',
    'GPTHC_SafeGemma_Train':'SafeGemma_GPTHC',
    'GPTHC_SafePhi':'SafePhi',
    'GPTHC_SafePhiFT':'SafePhi_GPTHC'

}

# # Apply the mapping to the group column
combined_df['source_file'] = combined_df['source_file'].replace(model_mapping)

filter=[ 'WildGuard', 'OpenAI','LlamaGuard','SafePhi','SafePhi_GPTHC']
df=combined_df[(combined_df['class'] == 'macro') & (combined_df['source_file'].isin(filter))]
df['file']=df['file'].str.capitalize()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def lineplot(df, score):
    plt.figure(figsize=(10, 6))

    # Set the font style for the entire plot
    plt.rcParams['font.family'] = 'Sans-serif'  # Use serif font (e.g., Times New Roman)
    plt.rcParams['font.weight'] = 'normal'  # Normal or bold
    plt.rcParams['font.size'] = 14  # Base font size

    # Cold-themed palette
    cold_palette = sns.color_palette("colorblind", as_cmap=False)

    # Lineplot with cold colors
    sns.lineplot(data=df, x='file', y=score, hue='source_file', marker='o', palette=cold_palette)

    # Highlight SafePhi outperforming instances
    highlight_df = df[df['source_file'] == 'SafePhi']
    for index, row in highlight_df.iterrows():
        if row[score] > df[(df['file'] == row['file']) & (df['source_file'] == 'SafePhi')][score].max():
            plt.scatter(
                x=row['file'],
                y=row[score],
                color='red',
                s=100,
                label="SafePhi Outperforms" if 'SafePhi Outperforms' not in plt.gca().get_legend_handles_labels()[1] else "",
                zorder=5
            )

    # Customize labels and ticks
    plt.xlabel("", fontsize=12, rotation=0)  # Bold x-axis label
    plt.ylabel("Macro F1", fontsize=15)  # Bold y-axis label
    plt.xticks(rotation=45, fontsize=15)  # Rotate x-axis labels
    plt.yticks(fontsize=12)  # Set y-axis tick font size

    # Add grid
    plt.grid(axis='y', linestyle='--', alpha=0.4)

    # Adjust legend
    handles, labels = plt.gca().get_legend_handles_labels()
    plt.legend(
        handles=handles,
        labels=labels,
        title='',
        loc='lower right',  # Anchor point for the legend
        # bbox_to_anchor=(-0.1, -0.1),  # Move legend outside the plot to the right
        ncol=1,  # Single column for vertical orientation
        fontsize=16,
        prop={'family': 'Sans-serif', 'size': 14},
        borderaxespad=0.  # Remove padding between plot and legend
    )

    # Save the plot
    plt.tight_layout()
    plt.savefig(f'/home/naseem_fordham/LLM_evaluation/Moderators_results/plots/GPTHC_{score}.png', dpi=600, bbox_inches='tight')

# Call the function
lineplot(df.round(2), 'f1_score')
```
